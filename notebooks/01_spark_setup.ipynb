## Vérification que les Workers exécutent bien du code Python
Un RDD force l'exécution de Python sur les executors (même version de Python que le driver, grâce à l'image commune pedali-spark).

```python
import platform
print('Python executors :', set(sc.parallelize(range(8), 4).map(lambda _: platform.python_version()).collect()))
print('Somme 0..999999 :', sc.parallelize(range(1_000_000), 8).sum())
```
